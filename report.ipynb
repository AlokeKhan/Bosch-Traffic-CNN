In this project, we trained a convolutional neural network (CNN) to classify images of traffic lights into three categories: Red, Yellow, and Green. The goal is to evaluate model performance on this multiclass classification task and understand the effectiveness of CNNs for visual traffic signal recognition.

We used a labeled dataset containing images of traffic lights in three categories. The training set contains approximately 13,000+ images, and the test set contains around 13,000 images as well. The dataset was preprocessed into fixed-size 32x32 RGB images.


The CNN model consists of two convolutional blocks followed by a fully connected layer with dropout. We used ReLU activations in intermediate layers and softmax in the final output layer for multiclass classification.
